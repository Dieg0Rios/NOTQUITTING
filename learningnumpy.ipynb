```python
import numpy as np
```

np.array is used to initialize the array of values

np.ndarray is used to make a randomly innitialized array of SHAPE 1 by 2. 

To access an element in a numpy array you can do it by: normal indexing, slice indexing  SLICING: [START : STOP : STEP]



```python
a = np.array([[1,2],[3,6],[4,5]])
print("This is a:", a)
print("This is the first element in A:", a[0]) 
```

```python
b = np.ndarray([2,2])
print("This is b:", b)
print("This is the shape of b:", b.shape)

```

```python
string = "THISISINSANE"
#         01234567891011
print(string[6:])
```

To get an element from a Two and higher-dimensional arrays its like so: 

```python
c = np.array([[1,2],[3,6],[4,5]])
print(c)
# IF we want to get the first column of c, we can do this:
print(c[:,0])
# The structure of it is: array[row, column]

# So to get 2: 
print(c[0,1]) #(first row, second column)
```
